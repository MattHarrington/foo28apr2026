```python
import io
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display

rng = np.random.default_rng(0)
```

# Sample plots

A few small matplotlib figures to show typical notebook output.

```python
x = np.linspace(0, 4 * np.pi, 1000)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.set_title("Trig functions")
ax.legend()
plt.show()
```

```python
data = rng.normal(size=(5000, 3))
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(data, bins=60, label=["a", "b", "c"], alpha=0.7)
ax.set_title("Three normal distributions")
ax.legend()
plt.show()
```

# Large random images

These cells emit several large PNGs of random pixel data. Random data does not
compress well, so each PNG stays close to its raw byte size. The combined
base64-encoded outputs push the notebook to roughly 50 MB on disk.

```python
def random_png(size: int) -> bytes:
    arr = rng.integers(0, 256, size=(size, size, 3), dtype=np.uint8)
    buf = io.BytesIO()
    Image.fromarray(arr).save(buf, format="PNG", compress_level=0)
    return buf.getvalue()

for i in range(7):
    png = random_png(1450)
    print(f"image {i}: {len(png) / 1e6:.2f} MB")
    display({"image/png": png}, raw=True)
```
